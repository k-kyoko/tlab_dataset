このノートブックは Docker 上の JupyterLab で `work/magnitude_limit.py` を動かすための最小サンプルです。

- `exact_limit(D)` : 有理数距離に対して lim 1^T A(t)^+ 1 を厳密に計算
- `schur_value(D, t)` : 任意の数値距離に対して高精度に評価
- `schur_samples(D)` : t を振って収束の様子を見る

```python
import sys, pathlib

# ノートブックをどこに置いても work/ を import できるようにする
here = pathlib.Path.cwd()
for candidate in (here, here / "work", *here.parents):
    if (candidate / "magnitude_limit.py").exists():
        sys.path.insert(0, str(candidate))
        break

from genmag_exact import magnitude_limit as ml
print(ml.__file__)
```

```python
# 例: 3 点のパスグラフ距離
D = [
    [0, 1, 2],
    [1, 0, 1],
    [2, 1, 0],
]

result = ml.exact_limit(D)
print("limit          :", result.limit)
print("method         :", result.method)
print("x definition   :", result.x_definition)
print("generic rank   :", result.generic_rank)
print("rank at inf    :", result.rank_at_infinity)
```

```python
# 数値的な確認: t を大きくしていったときの値
print(ml.format_samples(ml.schur_samples(D, p_values=(2, 4, 6, 8))))
```

```python
# 1 点だけ評価したい場合
v = ml.schur_value(D, t=10, dps=80)
print("t              :", v.t)
print("total          :", v.total)
print("range part     :", v.range_part)
print("kernel corr.   :", v.kernel_correction)
```
